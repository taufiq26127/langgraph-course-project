```python
from typing import TypedDict, List, Optional
from typing_extensions import Annotated
from pydantic import BaseModel, Field

class FullName(BaseModel):
    first_name: str
    last_name: str


class Student(BaseModel):
    name: Annotated[FullName, Field(... , description="The full name of the student")]
    age: Annotated[int, Field(..., gt=17, lt=25 ,description="The age of the student")]
    eligible: Annotated[bool, Field(default=None, description="Whether the student is eligible for the program")]
    hobbies: Annotated[List[str], Field(..., description="A list of hobbies of the student")]
    best_number: Annotated[Optional[List[int]], Field(default=None, max_length=3, description="A list of the student's best numbers, with a maximum length of 3")]


def show_student(student: Student):
    print(f"Student Name: {student.name.first_name} {student.name.last_name}")
    print(f"Age: {student.age}")
    print(f"Eligible: {student.eligible}")
    print(f"Hobbies: {', '.join(student.hobbies)}")
    print("======================================")
    print("data has been validated and is correct")

student_data = {
    "name": {
        "first_name": "John",
        "last_name": "Doe"
    },
    "age": 19,
    "hobbies": ["reading", "swimming", "coding"],
    "best_number": [7, 42, 3]
}

student1 = Student(**student_data)

show_student(student1)
```

```python
from pydantic import BaseModel

class Address(BaseModel):

    city: str
    state: str
    pin: str

class PatientData(BaseModel):

    name: str
    gender: str
    age: int
    address: Address

address_dict = {'city': 'gurgaon', 'state': 'haryana', 'pin': '122001'}

address1 = Address(**address_dict)

patient_dict = {'name': 'Kishor', 'gender': 'male', 'age': 40, 'address': address1}

patient1 = PatientData(**patient_dict)
print(patient1)
print(type(patient1))
```

```python
from typing import Annotated, List, TypedDict, Optional
from langgraph.graph.message import add_messages
from langgraph.graph import StateGraph, START, END
from langchain_core.messages import HumanMessage, AIMessage
from pydantic import BaseModel
import json

# ============================================
# 1. DEFINISIKAN STRUCTURED OUTPUT (Pydantic)
# ============================================
class ContactInfo(BaseModel):
    nama: str
    email: str
    telepon: str


# ============================================
# 2. STATE
# ============================================
class State(TypedDict):
    messages: Annotated[List, add_messages]
    contact: Optional[ContactInfo]


# ============================================
# 3. FAKE LLM (simulasi, biar tidak butuh API key)
#    -- di real project, ganti ini pakai model.with_structured_output(ContactInfo)
# ============================================
class FakeLLM:
    def invoke(self, messages):
        teks = messages[-1].content
        # simulasi ekstraksi sederhana (real-nya LLM yang lakukan ini)
        return ContactInfo(nama="Budi Santoso", email="budi@mail.com", telepon="081234567890")

llm = FakeLLM()


# ============================================
# 4. NODE -- ekstrak data, simpan ke state
# ============================================
def extract_contact(state: State):
    hasil = llm.invoke(state["messages"])
    ai_msg = AIMessage(content=f"Berhasil ekstrak: {hasil.nama}")
    return {
        "messages": [ai_msg],
        "contact": hasil
    }


# ============================================
# 5. BANGUN GRAPH
# ============================================
graph_builder = StateGraph(State)
graph_builder.add_node("extract", extract_contact)
graph_builder.add_edge(START, "extract")
graph_builder.add_edge("extract", END)
graph = graph_builder.compile()


# ============================================
# 6. JALANKAN
# ============================================
teks_input = "Nama saya Budi Santoso, email budi@mail.com, telp 081234567890"
result = graph.invoke({
    "messages": [HumanMessage(teks_input)],
    "contact": None
})

print("=== HASIL EKSTRAKSI ===")
print(result)
contact = result["contact"]
print(f"Nama    : {contact.nama}")
print(f"Email   : {contact.email}")
print(f"Telepon : {contact.telepon}")


# ============================================
# 7. SIMPAN HASIL KE FILE JSON -- di sinilah model_dump_json() dipakai
# ============================================
print()
print("=== SIMPAN KE FILE JSON ===")

with open("hasil_kontak.json", "w") as f:
    f.write(contact.model_dump_json(indent=2))

print("Tersimpan ke hasil_kontak.json, isinya:")
with open("hasil_kontak.json") as f:
    print(f.read())


# ============================================
# 8. KALAU MAU KIRIM KE API (simulasi, tanpa API beneran)
# ============================================
print()
print("=== SIMULASI KIRIM KE API ===")
print("requests.post('https://api.contoh.com/kontak', data=contact.model_dump_json())")
print("Data yang dikirim:", contact.model_dump_json())
```
